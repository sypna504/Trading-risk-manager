```python
from preparing_data import *
import pandas as pd
from pathlib import Path
```

```python
limit = 1000
downloader = GetCandles(interval="1h", limit=limit)

parquet_path = Path("history_data.parquet")
if parquet_path.exists():
    old_df = pd.read_parquet(parquet_path)
else:
    old_df = pd.DataFrame()
new_df = pd.DataFrame()
pairs = [
    "BTCUSDT", "ETHUSDT", "BNBUSDT", "USDCUSDT", "XRPUSDT", "SOLUSDT", "ADAUSDT", "DOGEUSDT", "TRXUSDT", "AVAXUSDT",
    "DOTUSDT", "LINKUSDT", "TONUSDT", "SHIBUSDT", "MATICUSDT", "LTCUSDT", "DAIUSDT", "BCHUSDT", "UNIUSDT", "ICPUSDT",
    "NEARUSDT", "APTUSDT", "FILUSDT", "XLMUSDT", "STXUSDT", "IMXUSDT", "GRTUSDT", "RENDERUSDT", "OPUSDT",
    "SUIUSDT", "ATOMUSDT", "AAVEUSDT", "THETAUSDT", "LDOUSDT", "ALGOUSDT", "FETUSDT", "VETUSDT", "EGLDUSDT", "TIAUSDT",
    "MKRUSDT", "FLOWUSDT", "GALAUSDT", "SANDUSDT", "MANAUSDT", "QNTUSDT", "CHZUSDT", "MINAUSDT", "PEPEUSDT"
]
bad_pairs = []
for pair in pairs:
    try:
        downloader.symbol = pair
        bi = downloader.get_binance_candles().to_dataframe()
        bi["symbol"] = pair
        # display(bi.head())

        new_df = pd.concat([new_df, bi], ignore_index=True)

        print(f"count of {pair}={len(bi)}")
        time.sleep(0.5)

    except Exception as e:
        print(f"skip {pair}: {e}")
        bad_pairs.append(pair)
        time.sleep(1)

df = pd.concat([old_df, new_df], ignore_index=True)
df = df.drop_duplicates(subset=["symbol", "timestamp"], keep="last")
if len(bad_pairs)!=0: print(bad_pairs) 
```

```python
new_df["timestamp"] = pd.to_datetime(new_df["timestamp"])
if not old_df.empty:
    old_df["timestamp"] = pd.to_datetime(old_df["timestamp"])
print("самая последняя дата из old_df=",old_df["timestamp"].max())
print("самая первая дата из new_df=", new_df["timestamp"].min())
date_check = (
    df
    .groupby("symbol")
    .agg(
        rows=("timestamp", "count"),
        min_date=("timestamp", "min"),
        max_date=("timestamp", "max"),
    )
    .sort_values("rows")
)

display(date_check)
```
