```python
limit = 5000
downloader = GetCandles(interval="1h", limit=limit)
df = pd.DataFrame()
pairs = [
    "BTCUSDT", "ETHUSDT", "BNBUSDT", "USDCUSDT", "XRPUSDT", "SOLUSDT", "ADAUSDT", "DOGEUSDT", "TRXUSDT", "AVAXUSDT",
    "DOTUSDT", "LINKUSDT", "TONUSDT", "SHIBUSDT", "MATICUSDT", "LTCUSDT", "DAIUSDT", "BCHUSDT", "UNIUSDT", "ICPUSDT",
    "NEARUSDT", "APTUSDT", "FILUSDT", "XLMUSDT", "STXUSDT", "IMXUSDT", "GRTUSDT", "RENDERUSDT", "OPUSDT",
    "SUIUSDT", "ATOMUSDT", "AAVEUSDT", "THETAUSDT", "LDOUSDT", "ALGOUSDT", "FETUSDT", "VETUSDT", "EGLDUSDT", "TIAUSDT",
    "MKRUSDT", "FLOWUSDT", "GALAUSDT", "SANDUSDT", "MANAUSDT", "QNTUSDT", "CHZUSDT", "MINAUSDT", "PEPEUSDT"
]
bad_pairs = []
for pair in pairs:
    try:
        downloader.symbol = pair
        bi = downloader.get_binance_candles().to_dataframe()
        bi["symbol"] = pair
        # display(bi.head())

        df = pd.concat([df, bi], ignore_index=True)

        print(f"count of {pair}={len(bi)}")
        time.sleep(0.5)

    except Exception as e:
        print(f"skip {pair}: {e}")
        bad_pairs.append(pair)
        time.sleep(1)
    
    

if len(bad_pairs)!=0: print(bad_pairs) 
```
